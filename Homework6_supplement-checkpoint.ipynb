```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    """Sigmoid activation function used in logistic regression."""
    return 1 / (1 + np.exp(-z))


# Data Generation 
np.random.seed(595)  # Keep dataset reproducible

# Generate random 2D feature matrix (2 × 100)
X = np.random.rand(2, 100)

# True model parameters (used to generate labels)
w_true = np.array([1.5, -2.5])
b_true = 1.0

# Compute true probabilities using the logistic model
probabilities = sigmoid(np.dot(w_true.T, X) + b_true)

# Convert probabilities to binary labels
Y = (probabilities > 0.5).astype(int)

# Add small noise to features for more realistic data
X = X + 0.3 * np.random.rand(2, 100) - 0.1 * np.random.rand(2, 100)

#  Train/Test Split 
X_train, X_test = X[:, :80], X[:, 80:]
Y_train, Y_test = Y[:80], Y[80:]

#  Parameter Initialization 
w = np.zeros(X.shape[0])   # Initialize weights to zero
b = 0.0                    # Initialize bias to zero

learning_rate = 5.95       # Tuned learning rate
num_epochs = 595           # Tuned number of iterations


#  Gradient Descent Training 
for epoch in range(num_epochs):

    # Compute model predictions for the training set
    A_train = sigmoid(np.dot(w.T, X_train) + b)

    # Compute gradients of cost J w.r.t. w and b
    dJdw = np.dot(X_train, (A_train - Y_train).T) / len(Y_train)   # Gradient for weights
    dJdb = np.mean(A_train - Y_train)                              # Gradient for bias

    # Update parameters
    w -= learning_rate * dJdw
    b -= learning_rate * dJdb


# Training & Testing Accuracy 
# Predictions on training set
A_train = sigmoid(np.dot(w.T, X_train) + b)
predictions_train = (A_train > 0.5).astype(int)

# Predictions on testing set
A_test = sigmoid(np.dot(w.T, X_test) + b)
predictions_test = (A_test > 0.5).astype(int)

# Compute accuracy
train_accuracy = np.mean(predictions_train == Y_train)
test_accuracy = np.mean(predictions_test == Y_test)

print(f"Training Set Accuracy: {train_accuracy:.2f}")
print(f"Test   Set Accuracy: {test_accuracy:.2f}")
```

```python
# Scatter plot of training data colored by their label
plt.scatter(X[0, :80], X[1, :80], c=Y[:80], cmap=plt.cm.Paired)

# Set axis limits for better visualization
plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)

ax = plt.gca()

# Generate a grid of points to evaluate the decision boundary
xlim = ax.get_xlim()
ylim = ax.get_ylim()
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 50),
                     np.linspace(ylim[0], ylim[1], 50))

# Evaluate logistic regression model on the grid
Z = np.dot(w.T, np.c_[xx.ravel(), yy.ravel()].T) + b
Z = sigmoid(Z)
Z = Z.reshape(xx.shape)

# Contour line at probability = 0.5 (decision boundary)
plt.contour(xx, yy, Z, colors='k', levels=[0.5], alpha=0.5, linestyles=['--'])

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Logistic Regression Decision Boundary')
plt.show()
```

```python
# Scatter plot of testing data (colored by true labels)
# Here we visualize the model's performance on the test set
plt.scatter(X[0, 80:], X[1, 80:], c=Y[80:], cmap=plt.cm.Paired)

# Set display limits for a consistent plot area
plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)

# Get the current axis object for extracting x/y limits
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# Create a 2D grid to evaluate the logistic model over the plane
xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 50),
                     np.linspace(ylim[0], ylim[1], 50))

# Compute model output (probabilities) on the grid
Z = np.dot(w.T, np.c_[xx.ravel(), yy.ravel()].T) + b
Z = sigmoid(Z)
Z = Z.reshape(xx.shape)

# Draw the decision boundary at probability = 0.5
# This contour line visually separates the two predicted classes
plt.contour(xx, yy, Z, colors='k', levels=[0.5], alpha=0.5, linestyles=['--'])

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Logistic Regression Decision Boundary (Test Set)')
plt.show()
```
