## 2. Implementing Gradient Descent Algorithm (60 pts)

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function f(x).
def f(x):
    return np.sqrt(x**2 + 5)

# Plot the function f(x), np.linspace might be helpful.
x_vals = np.linspace(-5, 5, 400)
y_vals = f(x_vals)

plt.figure(figsize=(6,4))
plt.plot(x_vals, y_vals)
plt.title("Function f(x) = sqrt(x^2 + 5)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()

```

```python
# Define the derivative of the function f(x), for gradient descent
def df(x):
    return x / np.sqrt(x**2 + 5)

# write a function to run the Gradient Descent Algorithm,
# take input arguments x_0, step_size, and num_iterations;
# return a list/array of x_k values.
def gradient_descent(x_0, step_size, num_iterations):
    x_list = [x_0]
    x = x_0
    for _ in range(num_iterations):
        x = x - step_size * df(x)
        x_list.append(x)
    return np.array(x_list)

# set up/assign initial values
last_digit = 2   # the last digit of my SB ID 
x_0 = last_digit / 5 + 1
step_size = 1
num_iterations = 50

# Run gradient descent by calling the function
x_list = gradient_descent(x_0, step_size, num_iterations)

# report x_T by printing out the value
print("Final x_T =", x_list[-1])

# Plot the f(x_k) values vs. iteration number
f_vals = f(x_list)

plt.figure(figsize=(6,4))
plt.plot(range(len(f_vals)), f_vals)
plt.title("f(x_k) vs Iteration Number (step size = 1)")
plt.xlabel("Iteration k")
plt.ylabel("f(x_k)")
plt.grid(True)
plt.show()

```

```python
# create a list containing step sizes
step_sizes = [5, 3, 1, 0.5]

# initialize a list to store the results from different step sizes
results = []

# loop through the step sizes and perform GD and record results
for alpha in step_sizes:
    x_list = gradient_descent(x_0, alpha, num_iterations)
    f_vals = f(x_list)
    diff_vals = f_vals[:-1] - f_vals[1:]     # f(x_{k-1}) - f(x_k)
    results.append((alpha, diff_vals))

# plot multiple curves in a single graph
plt.figure(figsize=(7,5))

for alpha, diff_vals in results:
    plt.plot(diff_vals, label=f"step size = {alpha}")

plt.title("f(x_{k-1}) - f(x_k) for Different Step Sizes")
plt.xlabel("Iteration k")
plt.ylabel("Decrease in f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

## 4. Logistic Regression with Numpy and Graphing with Matplotlib (20 pts)

In class, we implemented a logistic regression model to perform binary classification. Here, I provided the exact same code that we discussed in class, but the values for learning_rate and num_epochs are modified. I also removed all the comments. Your task is to

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the logistic (sigmoid) function
def sigmoid(z):
   
    # Compute logistic sigmoid of input z.
    
    return 1 / (1 + np.exp(-z))

# 2. Generate synthetic binary-classification data
np.random.seed(42)  #  SB ID 
N = 200
# 2D features (feature1, feature2)
X = np.random.randn(N, 2)
# Construct labels: e.g. a linear separation plus noise
# Here: class 1 if w_true · x + b_true + noise > 0, else class 0
w_true = np.array([2.0, -1.0])
b_true = 0.3
noise = 0.5 * np.random.randn(N)
y = (X.dot(w_true) + b_true + noise > 0).astype(int)

# Optional: split into train / test
shuffled_idx = np.random.permutation(N)
train_frac = 0.8
n_train = int(train_frac * N)
train_idx = shuffled_idx[:n_train]
test_idx = shuffled_idx[n_train:]

X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]

# 3. Set hyperparameters and initialize parameters (weights and bias)
lr = 0.1         # learning rate
num_epochs = 500 # epochs (iterations)
m, d = X_train.shape
w = np.zeros(d)
b = 0.0

# 4. Training (gradient descent)
for epoch in range(num_epochs):
    z = X_train.dot(w) + b
    y_hat = sigmoid(z)                     # predicted probabilities
    error = y_hat - y_train

    # compute gradient
    grad_w = (1 / m) * X_train.T.dot(error)
    grad_b = (1 / m) * np.sum(error)

    # update parameters
    w -= lr * grad_w
    b -= lr * grad_b

# 5. Compute training and testing accuracy
def predict(X, w, b, threshold=0.5):
    probs = sigmoid(X.dot(w) + b)
    return (probs >= threshold).astype(int)

y_train_pred = predict(X_train, w, b)
y_test_pred = predict(X_test, w, b)

train_acc = np.mean(y_train_pred == y_train)
test_acc = np.mean(y_test_pred == y_test)

print(f"Training accuracy: {train_acc*100:.2f}%")
print(f"Test accuracy:     {test_acc*100:.2f}%")
print("Learned parameters: w =", w, ", b =", b)

# 6. Plot the training dataset and the decision boundary
plt.figure(figsize=(6,6))
plt.scatter(X_train[y_train==0,0], X_train[y_train==0,1], c='red', label='Class 0')
plt.scatter(X_train[y_train==1,0], X_train[y_train==1,1], c='blue', label='Class 1')

# Create a grid to evaluate model
xmin, xmax = X[:,0].min() - 1, X[:,0].max() + 1
ymin, ymax = X[:,1].min() - 1, X[:,1].max() + 1
xx, yy = np.meshgrid(np.linspace(xmin, xmax, 200),
                     np.linspace(ymin, ymax, 200))
grid = np.c_[xx.ravel(), yy.ravel()]
probs = sigmoid(grid.dot(w) + b).reshape(xx.shape)

# plot decision boundary (contour at p = 0.5)
plt.contour(xx, yy, probs, levels=[0.5], linewidths=2, colors='k')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Logistic Regression — Data & Decision Boundary")
plt.legend()
plt.show()
```

```python
## from the hint code:
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    """Sigmoid activation function used in logistic regression."""
    return 1 / (1 + np.exp(-z))


# Data Generation 
np.random.seed(595)  # Keep dataset reproducible

# Generate random 2D feature matrix (2 × 100)
X = np.random.rand(2, 100)

# True model parameters (used to generate labels)
w_true = np.array([1.5, -2.5])
b_true = 1.0

# Compute true probabilities using the logistic model
probabilities = sigmoid(np.dot(w_true.T, X) + b_true)

# Convert probabilities to binary labels
Y = (probabilities > 0.5).astype(int)

# Add small noise to features for more realistic data
X = X + 0.3 * np.random.rand(2, 100) - 0.1 * np.random.rand(2, 100)

#  Train/Test Split 
X_train, X_test = X[:, :80], X[:, 80:]
Y_train, Y_test = Y[:80], Y[80:]

#  Parameter Initialization 
w = np.zeros(X.shape[0])   # Initialize weights to zero
b = 0.0                    # Initialize bias to zero

learning_rate = 5.95       # Tuned learning rate
num_epochs = 595           # Tuned number of iterations


#  Gradient Descent Training 
for epoch in range(num_epochs):

    # Compute model predictions for the training set
    A_train = sigmoid(np.dot(w.T, X_train) + b)

    # Compute gradients of cost J w.r.t. w and b
    dJdw = np.dot(X_train, (A_train - Y_train).T) / len(Y_train)   # Gradient for weights
    dJdb = np.mean(A_train - Y_train)                              # Gradient for bias

    # Update parameters
    w -= learning_rate * dJdw
    b -= learning_rate * dJdb


# Training & Testing Accuracy 
# Predictions on training set
A_train = sigmoid(np.dot(w.T, X_train) + b)
predictions_train = (A_train > 0.5).astype(int)

# Predictions on testing set
A_test = sigmoid(np.dot(w.T, X_test) + b)
predictions_test = (A_test > 0.5).astype(int)

# Compute accuracy
train_accuracy = np.mean(predictions_train == Y_train)
test_accuracy = np.mean(predictions_test == Y_test)

print(f"Training Set Accuracy: {train_accuracy:.2f}")
print(f"Test   Set Accuracy: {test_accuracy:.2f}")
```
